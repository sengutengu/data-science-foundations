# Data Visualization with Matplotlib and Seaborn

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(5, 100, 100)
y = np.linspace(10, 1000, 100)

plt.plot(x,y)

plt.xlabel("Current")
plt.ylabel("Voltage")
plt.title("Ohm's Law")

plt.xlim([20, 80]) # limit range shown in x and y axes
plt.ylim([200, 800])

plt.show() # required to explicitly show plot
```

```python
plt.plot(x,y, 'b+') # blue plus-signs

plt.xlabel("Current")
plt.ylabel("Voltage")
plt.title("Ohm's Law")

plt.xlim([20, 80])
plt.ylim([200, 800])

plt.show()
```

```python
x = np.linspace(1, 10, 10)
y = np.linspace(1, 10, 10)

plt.plot(x, y, 'r-')
plt.plot(x, y**2, 'b+')
plt.plot(x, y**3, 'g^') # multiple options

plt.show()
```

```python
plt.plot([1,4,6,8],[3,8,3,5]) # also works with lists but converted into np array internally
plt.show()
```

## Figures and Subplots

```python
x = np.linspace(1, 10, 100)
y = np.log(x)

plt.figure(1) # initiate a new figure

plt.subplot(1,2,1) # one row, two columns, subplot 1
plt.title("y = log(x)")
plt.plot(x, y)

plt.subplot(1,2,2) # one row, two columns, subplot 2
plt.title("y = log(x)^2")
plt.plot(x, y**2)

plt.show()
```

## Seaborn

```python
import seaborn as sns # built on matplotlib, allows more advanced functionalities
```

```python
sns.set_style('whitegrid')

sns.histplot(df['Shipping_Cost'], kde=True)
sns.rugplot(df['Shipping_Cost']) # rugplot shows where the data is on x axis
plt.show()
```

```python
sns.displot(df['Shipping_Cost'])
plt.show()
```

```python
sns.boxplot(df['Order_Quantity'])
plt.show()
```

```python
sns.boxplot(x=df['Order_Quantity']) # put on side if desired
plt.show()
```

```python
sns.jointplot(data=df, x='Sales', y='Profit')
plt.show()
```

```python
df = df[(df.Profit<10000) & (df.Sales < 20000)] # filter out outliers

sns.jointplot(data=df, x='Sales', y='Profit')
plt.show()
```

```python
df = df[(df.Profit<100) & (df.Profit>-100) & (df.Sales < 200)] # filter out outliers

sns.jointplot(data=df, x='Sales', y='Profit', kind='hex', color='k') # hex graph
plt.show()
```

```python
sns.pairplot(df)
```
