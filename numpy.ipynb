# Intro to Numpy

We use Numpy because it's fast and convenient. 

The basis object in numpy is `ndarray`, short for *n*-dimensional array. `ndarray`s must be homogeneous---that is, composed of the same data type.   

```python
import numpy as np
```

```python
a = np.array([1,2,3])
```

```python
a.shape
```
